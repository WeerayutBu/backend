# Mocking
Replace a slow external dependency and verify the interaction.


```python
from unittest.mock import AsyncMock

provider = AsyncMock(return_value="hello")
result = await provider("prompt")
provider.assert_awaited_once_with("prompt")
print(result)
```

## Polished version
Mock an explicit boundary while testing the service's behavior and call contract.


```python
from typing import Protocol

class CompletionProvider(Protocol):
    async def complete(self, prompt: str) -> str: ...

class CompletionService:
    def __init__(self, provider: CompletionProvider) -> None:
        self.provider = provider
    async def answer(self, prompt: str) -> str:
        return (await self.provider.complete(prompt)).strip()

mock_provider = AsyncMock(spec=CompletionProvider)
mock_provider.complete.return_value = "  hello  "
service = CompletionService(mock_provider)
assert await service.answer("Hi") == "hello"
mock_provider.complete.assert_awaited_once_with("Hi")
print("passed")
```
