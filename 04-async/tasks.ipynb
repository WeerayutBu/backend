# Async tasks
Start independent I/O operations together and wait for their results.


```python
import asyncio

async def work(value: int) -> int:
    await asyncio.sleep(0.01)
    return value * 2

task = asyncio.create_task(work(3))
print(await task)
```

## Polished version
Use structured concurrency so child tasks finish or fail as one unit.


```python
class BatchRunner:
    async def run(self, values: list[int]) -> list[int]:
        tasks: list[asyncio.Task[int]] = []
        async with asyncio.TaskGroup() as group:
            for value in values:
                tasks.append(group.create_task(work(value)))
        return [task.result() for task in tasks]

print(await BatchRunner().run([1, 2, 3]))
```
